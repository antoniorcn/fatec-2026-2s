```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
gerador_randomico = np.random.default_rng( 100 )
```

```python
# Variavel independente
criancas_idades = gerador_randomico.integers( 5, 18, 50 )
criancas_idades
```

```python
ratios = gerador_randomico.integers( 7, 13, 50 )
ratios
```

```python
# Variavel Dependente
criancas_alturas = criancas_idades * ratios
criancas_alturas
```

```python
w = 9
b = 6
```

```python
soma = 0
for idx, idade in enumerate(criancas_idades):
    y_hat = idade * w + b  # Criança Altura Estimada
    y = criancas_alturas[idx]
    erro = (y_hat - y) ** 2
    soma = soma + erro
m = len(criancas_idades)
custo = soma / (2 * m)
print(f"W:{w}\tB:{b}\tCusto:{custo}")
```

```python
def calcular_custo( X, Y, w, b ):
    soma = 0
    for idx, x in enumerate(X):
        y_hat = x * w + b  # Calcula o valor do y_hat baseado em uma instancia do X
        y = Y[idx]
        erro = (y_hat - y) ** 2
        soma = soma + erro
    m = len(X)
    custo = soma / (2 * m)
    # print(f"W:{w}\tB:{b}\tCusto:{custo}")
    return custo
```

```python
def calcular_derivada_w( X, Y, w, b ):
    soma = 0
    for idx, x in enumerate(X):
        y_hat = x * w + b  # Calcula o valor do y_hat baseado em uma instancia do X
        y = Y[idx]
        erro = (y_hat - y) * x
        soma = soma + erro
    m = len(X)
    dW = soma / m
    return dW
```

```python
def calcular_derivada_b( X, Y, w, b ):
    soma = 0
    for idx, x in enumerate(X):
        y_hat = x * w + b  # Calcula o valor do y_hat baseado em uma instancia do X
        y = Y[idx]
        erro = (y_hat - y)
        soma = soma + erro
    m = len(X)
    dB = soma / m
    return dB
```

```python
X = criancas_idades    ## [15,  8,  9,  9,  8,  5, 12 ....
Y = criancas_alturas   ## [105,  72,  90,  72,  64,  55 ... 
```

```python
custo = 100000000
ciclos = 150
custo_minimo = 100
ciclo_atual = 0
w = 10
b = 20
alpha = 0.001
while ciclo_atual < ciclos and custo > custo_minimo:
    custo = calcular_custo(X, Y, w, b)
    print(f"Custo: {custo}\tW:{w}\tB:{b}")
    dw = calcular_derivada_w(X, Y, w, b)
    db = calcular_derivada_b(X, Y, w, b)
    w = w - alpha * dw
    b = b - alpha * db
    ciclo_atual = ciclo_atual + 1 
    
    
```

### Graficos

```python
Y_hat = criancas_idades * w + b
```

```python
plt.scatter(criancas_idades, criancas_alturas)
plt.plot(criancas_idades, Y_hat, color="red")
plt.show()
```
