```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
gerador_randomico = np.random.default_rng( 100 )
```

```python
criancas_idades = gerador_randomico.integers( 5, 18, 50 )
criancas_idades
```

```python
ratios = gerador_randomico.integers( 7, 13, 50 )
ratios
```

```python
criancas_alturas = criancas_idades * ratios
criancas_alturas
```

```python
w = 10
b = 10
y_hat = criancas_idades * w + b
plt.scatter( criancas_idades, criancas_alturas )
plt.plot(criancas_idades, y_hat, color="red" )
plt.show()
```

```python
x1 = 11
y1 = 99
```

```python
y_hat1 = x1 * w + b
y_hat1
```

```python
erro1 = (y_hat1 - y1) ** 2
erro1
```

```python
x2 = 15
y2 = 180
y_hat2 = x2 * w + b
erro2 = (y_hat2 - y2) ** 2
print("Y_hat", y_hat2)
print("Erro", erro2)
```

```python
w = 9
b = 6
```

```python
soma = 0
for idx, idade in enumerate(criancas_idades):
    y_hat = idade * w + b  # Criança Altura Estimada
    y = criancas_alturas[idx]
    erro = (y_hat - y) ** 2
    soma = soma + erro
m = len(criancas_idades)
custo = soma / (2 * m)
print(f"W:{w}\tB:{b}\tCusto:{custo}")
```
