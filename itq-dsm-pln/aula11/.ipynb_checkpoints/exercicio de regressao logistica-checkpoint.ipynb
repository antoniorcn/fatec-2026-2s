```python
import math
```

```python
import pandas as pd
```

```python
euler = 2.71828
```

```python
dados = pd.DataFrame( {
    "X1": [3, 0, 6, 2], # Quantidade de palavras suspeitas
    "X2": [1, 2, 5, 3], # Quantidade de links
    "Y": [0, 0, 1, 1]   # Spam ou Não Spam
})
dados
```

```python
w1 = 5.0
w2 = 1.0
b = 3.0
```

```python
def sigmoid( z ):
    return 1.0 / (1.0 + (euler ** -z))
```

```python
def funcao_soma( x1, x2, w1, w2, b ):
    return x1 * w1 + x2 * w2 + b
```

```python
def erro_regressao_logistica(p_hat, y):
    erro = ( (y * math.log(p_hat)) +  ( (1-y)*math.log(max(1-p_hat, 0.00001))) )
    return erro
```

```python
ciclos_maximo = 30
ciclo_atual = 0
alpha = 0.01
J = 100000
while ciclo_atual < ciclos_maximo and abs(J) > 1:
    loss = 0
    m = dados.shape[0]
    soma_d_w1 = 0
    soma_d_w2 = 0
    soma_d_b = 0
    for i in range(m):
        x1 = dados.loc[i, "X1"]
        x2 = dados.loc[i, "X2"]
        y = dados.loc[i, "Y"]
        z = funcao_soma( x1, x2, w1, w2, b )
        p_hat = sigmoid( z )
        erro = erro_regressao_logistica(p_hat, y)
        loss = loss + erro
        # print(f"X1: {x1}\tX2:{x2}\tZ:{z}\tp_hat:{p_hat}\tErro:{erro}")
        soma_d_w1 = soma_d_w1 + ((p_hat - y) * x1)
        soma_d_w2 = soma_d_w2 + ((p_hat - y) * x2)
        soma_d_b = soma_d_b + (p_hat - y)
    d_w1 = (soma_d_w1 / m)
    d_w2 = (soma_d_w2 / m)
    d_b = (soma_d_b / m)
    J = - (loss / m)
    
    w1 = w1 - alpha * d_w1
    w2 = w2 - alpha * d_w2
    b = b - alpha * d_b
    print(f"Ciclo:{ciclo_atual}\tJ: {J}\tw1: {w1}\tw2: {w2}\tb: {b}")
    ciclo_atual = ciclo_atual + 1
    
    
```

```python
x1 = 3
x2 = 1
z = funcao_soma(x1, x2, w1, w2, b)
p_hat = sigmoid(z)
y_hat = 0
if p_hat >= 0.5: y_hat = 1
print("Spam: ", y_hat)
```
