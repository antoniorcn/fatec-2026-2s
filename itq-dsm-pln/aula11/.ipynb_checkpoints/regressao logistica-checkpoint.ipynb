```python
import numpy
import math
```

```python
euler = 2.71828
```

```python
# Valores de entrada
x1 = 3     # x_1  quantidade de palavras suspeitas
x2 = 2     # x_2  quantidade de links
y = 1      # saida esperada
```

```python
# Pesos
w1 = 0.8   # Peso do primeiro parametro
w2 = 0.5   # Peso do segundo parametro
b = -2     # Viez ou Bias
```

```python
z = w1 * x1 + w2 * x2 + b
z
```

```python
def sigmoid( z ):
    return 1.0 / (1.0 + (euler ** -z))
```

```python
p_hat = sigmoid(z)
p_hat
```

```python
limiar = 0.5
```

```python
y_hat = 0
if p_hat >= limiar:
    y_hat = 1
```

```python
y_hat
```

```python
# erro = y_hat - y  # Esta não é uma boa maneira de calcular o erro
erro = y - p_hat
erro
```

```python
p_hat_linha = sigmoid(z) * (1 - sigmoid(z))
p_hat_linha
```

```python
p_hat_linha = 0.0001 * (1 - 0.0001)
print(f"{p_hat_linha:.4f}")
```

```python
p_hat_linha =0.999 * (1 - 0.999)
print(f"{p_hat_linha:.4f}")
```

```python
erro = - ( (y * math.log(p_hat)) +  ( (1-y)*math.log(1-p_hat)) )
erro
```
