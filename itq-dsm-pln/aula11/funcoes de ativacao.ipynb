```python
import numpy
```

```python
euler = 2.71828
```

```python
# Valores de entrada
x1 = 3     # x_1  quantidade de palavras suspeitas
x2 = 2     # x_2  quantidade de links
```

```python
# Pesos
w1 = 0.8   # Peso do primeiro parametro
w2 = 0.5   # Peso do segundo parametro
b = -2     # Viez ou Bias
```

```python
x = w1 * x1 + w2 * x2 + b
x
```

```python
def threshold( z ):
    limiar = 0
    if z > 0:
        return 1
    else:
        return 0
```

```python
threshold( x )
```

```python
def sigmoid( z ):
    return 1.0 / (1.0 + (euler ** -z))
```

```python
sigmoid(x)
```

```python
def tanh( z ):
    return ((euler ** 2.0*z) - 1.0) / ((euler ** 2.0*z) + 1.0)
```

```python
tanh(x)
```

```python
tanh(0)
```
