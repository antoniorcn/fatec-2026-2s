```python
import spacy
```

```python
import matplotlib.pyplot as plt
```

```python
nlp = spacy.load("pt_core_news_sm")
```

```python
nome_arquivo = "C:\\git\\dados\\nlp\\machado-assis-memorias-postumas-braz-cubas.txt"
conteudo_arquivo = []
with open( nome_arquivo, "r", encoding="utf-8" ) as arquivo:
    conteudo_arquivo = arquivo.readlines()
```

```python
textao = " ".join(conteudo_arquivo)
```

```python
docs = nlp(textao)
```

```python
dicionario_lemma = []
frequencias = {}
```

```python
for token in docs:
    if token.lemma_ not in dicionario_lemma:
        dicionario_lemma.append(token.lemma_)
    frequencias[token.lemma_] = frequencias.get(token.lemma_, 0) + 1
```

```python
lista_frequencias = list(frequencias.items())
```

```python
lista_frequencias.sort(key=lambda x: x[1], reverse=True)
```

```python
lista_frequencias[:10]
```

```python
textos = [el[0] for el in lista_frequencias[:10]]
quantidades = [el[1] for el in lista_frequencias[:10]]
```
