```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
X = np.arange(0, 100)
X.shape
```

```python
aleatorios = np.random.random(100)
aleatorios.shape
```

```python
Y = X * aleatorios
Y.shape
```

```python
plt.plot(X, Y)
plt.show()
```

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].plot(X, Y, color="green", label="Compras")
axes[0, 0].plot(X, Y*1.5, color="purple", label="Vendas")
axes[0, 0].set_title("Grafico 1")
axes[0, 0].set_ylabel("Vendas")
axes[0, 0].set_xlabel("Ultimos 100 meses")
axes[0, 0].grid(visible=True, which="both", linestyle="--", color="yellow", alpha=0.2, linewidth=1)
axes[0, 0].legend(loc="upper right")
axes[0, 1].plot(X, Y, color="red")
axes[0, 1].set_title("Grafico 2")
axes[0, 1].grid(visible=True, which="both", linestyle=":", color="blue", alpha=0.2, linewidth=2)
axes[0, 1].text(75, 5, "Pandemia")
axes[0, 1].set_xlim(60, 90)
axes[0, 1].set_ylim(0, 50)
axes[0, 1].axhline(30, linewidth=4)
axes[0, 1].axvline(65, linewidth=4, color="purple")
axes[1, 0].plot(X, Y, color="blue", markerfacecolor="red", markersize=6, marker="*")
axes[1, 0].set_title("Grafico 3")
axes[1, 0].grid(visible=True, which="both", linestyle="-.", color="red", alpha=0.2, linewidth=3)
axes[1, 1].bar(X, Y, color="yellow", label="Receitas")
axes[1, 1].bar(X, Y*-0.8, color="red", label="Despesas")
# axes[1, 1].barh(X, Y, color="yellow")
# axes[1, 1].barh(X, Y*-0.8, color="red")
axes[1, 1].set_title("Grafico 4")
axes[1, 1].grid(visible=True, which="both", linestyle="-", color="green", alpha=0.2, linewidth=4)
axes[1, 1].set_xlim(25, 50)
axes[1, 1].set_ylim(-60, 60)
axes[1, 1].legend()
# axes[1, 1].set_xlim(-60, 60)
# axes[1, 1].set_ylim(25, 50)
plt.tight_layout(pad=1)
plt.show()
```

```python
quantidade_textos = 30
aleatorios = np.random.random( quantidade_textos )
quantidade_palavras = ((aleatorios * 80) + 20).astype(np.int32)
quantidade_palavras
```
