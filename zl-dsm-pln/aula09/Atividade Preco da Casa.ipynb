```python
import numpy as np
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
tamanho_casa = np.array([150, 160, 155, 140 ,170, 165, 152])   # Variavel Independente
preco_casa = np.array([300000, 350000, 320000, 280000, 400000, 370000, 310000])   # Variavel Dependente
```

```python
w = 2200
b = 1000
```

```python
reta = tamanho_casa * w + b
reta
```

```python
plt.scatter(tamanho_casa, preco_casa)
plt.plot(tamanho_casa, reta, color="red")
plt.scatter(tamanho_casa, reta, color="red")
plt.show()
```

```python
soma_erros = 0
m = len(tamanho_casa)
for idx, tamanho in enumerate(tamanho_casa):
    y = preco_casa[idx]
    y_hat = tamanho * w + b
    erro = (y_hat - y) ** 2
    soma_erros = soma_erros + erro
custo = soma_erros / (2 * m)
custo
```
