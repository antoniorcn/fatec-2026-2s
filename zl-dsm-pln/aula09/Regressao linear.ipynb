```python
df = pd.DataFrame({"Idade": idade_criancas, "Altura": alturas_criancas})
df
```

```python
w = 6.3
b = 40
reta = df["Idade"] * w + b

idade = 18
y_hat = idade * w + b  # altura prevista

plt.scatter( df["Idade"], df["Altura"] )
plt.plot(df["Idade"], reta, color="red" )
plt.scatter( idade, y_hat, color="magenta")
plt.scatter( df["Idade"], reta, color="red")
plt.show()
```

```python
soma_erros = 0
for idx, idade in enumerate(df["Idade"]):
    y = df["Altura"][idx]
    y_hat = idade * w + b
    erro = (y_hat - y) ** 2
    # print(f"Indice: {idx}\tIdade:{idade}\tY: {y}\tY_hat: {y_hat}\tErro: {erro}")
    soma_erros = soma_erros + erro
soma_erros
```
