```python
alturas_criancas = idade_criancas * np.random.randint(4, 8, 300) + np.random.randint(38, 43, 300)
alturas_criancas
```

```python
df = pd.DataFrame({"Idade": idade_criancas, "Altura": alturas_criancas})
df
```

```python
w = 6.3
b = 40
reta = df["Idade"] * w + b

idade = 18
y_hat = idade * w + b  # altura prevista

plt.scatter( df["Idade"], df["Altura"] )
plt.plot(df["Idade"], reta, color="red" )
plt.scatter( idade, y_hat, color="magenta")
plt.scatter( df["Idade"], reta, color="red")
plt.show()
```

```python
def previsao( X, w, b ): 
    y_hat = X * w + b
    return y_hat
```

```python
def calcular_custo(X, Y, Y_hat):
    """
    X -> corresponde aos valores da variavel independente
    Y -> corresponde aos valores reais da variavel dependente
    Y_hat -> corresponde aos valores calculados da variavel dependente
    """
    custo = 0
    for indice, valor in enumerate(X):
        y = Y[indice]
        y_hat = Y_hat[indice]
        erro = (y_hat - y) ** 2
        custo = custo + erro
    return custo
```

```python
def derivadas_parciais(X, Y, Y_hat):
    """
        X -> corresponde aos valores da variavel independente
        Y -> corresponde aos valores reais da variavel dependente
        Y_hat -> corresponde aos valores calculados da variavel dependente
    """
    m = len(X)
    soma_dW = 0
    soma_dB = 0
    for indice, x in enumerate(X):
        y = Y[indice]
        y_hat = Y_hat[indice]
        soma_dW = soma_dW + ((y_hat - y) * x)
        doma_dB = soma_dB + (y_hat - y)
    dW = soma_dW / m
    dB = soma_dB / m
    return (dW, dB)
        
```

```python
ciclos_maximo = 1000
custo_minimo = 91719
custo = custo_minimo
ciclo_atual = 0
w = 100
b = 40
alpha = 0.001
```

```python
while custo >= custo_minimo and ciclo_atual < ciclos_maximo:
    print(f"Ciclo:{ciclo_atual}\tCusto: {custo}\tW: {w}\tB:{b}")
    Y_hat = previsao( df["Idade"], w, b )
    custo = calcular_custo(df["Idade"], df["Altura"], Y_hat)
    dW, dB = derivadas_parciais(df["Idade"], df["Altura"], Y_hat)
    w = w - (alpha * dW)
    b = b - (alpha * dB)
    ciclo_atual = ciclo_atual + 1
```

```python
idade = 18
y_hat = idade * w + b  # altura prevista
reta = df["Idade"] * w + b
plt.scatter( df["Idade"], df["Altura"] )
plt.plot(df["Idade"], reta, color="red" )
plt.scatter( idade, y_hat, color="magenta")
plt.scatter( df["Idade"], reta, color="red")
plt.show()
```
