```python
euler = 2.71828
```

```python
def funcao_threshold( z ):
    if z > 0:
        return 1
    else:
        return 0
```

```python
def funcao_sigmoid( z ):
    return 1.0 / (1.0 + euler**(-z))
```

```python
def funcao_tanh( z ):
    return (euler ** (2.0 * z) - 1.0) / (euler ** (2.0 * z) + 1.0) 
```

```python
def funcao_ReLU( z ):
    if z < 0:
        return 0
    else:
        return z
```

```python
z = -4
print(f"Threshold Z={z}\t\ty_hat={funcao_threshold( z )}") 
print(f"Sigmoid Z={z}\t\ty_hat={funcao_sigmoid( z ):f}") 
print(f"Tanh Z={z}\t\ty_hat={funcao_tanh( z ):f}") 
print(f"ReLU Z={z}\t\ty_hat={funcao_ReLU( z ):f}") 
```

```python
y_hat = funcao_sigmoid( -4 )
print(f"{y_hat:f}")
```
